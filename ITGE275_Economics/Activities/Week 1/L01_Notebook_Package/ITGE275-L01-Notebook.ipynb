### **1. Introduction: Economics as a System of Constraints**

Welcome to **ITGE275**. As computer science students, you are likely already familiar with the concept of **optimization**. 

* When you write an algorithm, you try to maximize efficiency (Speed, Memory) subject to hardware constraints (CPU cycles, RAM, Bandwidth).
* **Economics is exactly the same thing.**

**Economics** is the science of allocating limited resources to maximize utility (happiness) or profit. In this session, we will treat an "Economy" (specifically, a Tech Startup) as a system. We will define its inputs (**Factors of Production**), its constraints (**Scarcity**), and the cost of making decisions (**Opportunity Cost**).

### **2. Case Study: "The Startup's Dilemma"**

Imagine you are the CTO of a new startup, **"TechNova"**. You have limited resources:

1.  **Seed Fund:** $1,000,000 (Capital)
2.  **Team:** 5 Developers (Labor)
3.  **Time:** 40 hours per week per developer (Total 200 hours/week).

You have two competing priorities for your infrastructure:
1.  **Server Capacity (Infrastructure):** Building robust backend systems to handle traffic.
2.  **Customer Support (Service):** Building tools and chat-bots to help users.

**The Constraint:** You cannot have infinite amounts of both. Every hour a developer spends optimizing the server is an hour *not* spent improving the support chatbot. This is the core economic problem.

### **4. Concept: The Production Possibility Frontier (PPF)**

**Scarcity** is the fundamental limitation of resources. In our startup, the scarce resource is **Developer Hours**.

We can visualize this trade-off using the **Production Possibility Frontier (PPF)** curve. The PPF shows the maximum possible output combinations of two goods (Server Capacity vs. Support Capacity) given a fixed set of resources.

* **Points on the curve:** Efficient (using all developers).
* **Points inside the curve:** Inefficient (developers are idle).
* **Points outside the curve:** Impossible (requires more resources than available).

#### **The Math: Diminishing Returns**
In real economics (and CS), trade-offs are rarely 1-to-1. As you move more developers to servers, they might get tired or run into "diminishing returns" (e.g., adding a 10th developer to a project adds less value than the 1st).

We model this using a non-linear production function:
$$ Output = Efficiency \times \sqrt{Input} $$

```python
def production_function(resource_input, efficiency=10.0):
    """
    A simple production function modeling diminishing returns (square root).
    Input: Resource hours (Labor)
    Output: Units of value produced
    """
    # We use sqrt to model that the first few hours are very productive,
    # but later hours add less marginal value.
    return efficiency * np.sqrt(resource_input)

def plot_static_ppf(total_hours=200):
    """
    Plots a static Production Possibility Frontier.
    """
    # Create a range of possible allocations (0 to total_hours)
    server_hours = np.linspace(0, total_hours, 100)
    support_hours = total_hours - server_hours

    # Calculate Output for each allocation scenario
    # We assume 'Server' work is slightly harder (eff=10) than 'Support' (eff=12)
    server_output = production_function(server_hours, efficiency=10)
    support_output = production_function(support_hours, efficiency=12)

    # Plotting
    fig, ax = plt.subplots()
    ax.plot(server_output, support_output, color='blue', linewidth=3, label='PPF Curve')
    ax.fill_between(server_output, support_output, color='blue', alpha=0.1)
    
    ax.set_title(f'Production Possibility Frontier (Total Hours: {total_hours})', fontsize=14)
    ax.set_xlabel('Server Capacity Units', fontsize=12)
    ax.set_ylabel('Customer Support Units', fontsize=12)
    ax.grid(True, linestyle='--', alpha=0.7)
    ax.legend()
    
    # Annotations for zones
    ax.text(20, 20, 'Inefficient Zone\n(Unused Resources)', fontsize=10, color='gray')
    ax.text(100, 150, 'Impossible Zone\n(Scarcity Constraint)', fontsize=10, color='red')
    
    plt.show()

# Run the static plot
plot_static_ppf()
```

### **5. Interactive Lab: Opportunity Cost**

**Opportunity Cost** is the value of the *next best alternative* foregone.

* **CS Analogy:** If a CPU core is processing a background task, it cannot process the active window. The opportunity cost is the lag in the active window.
* **Economic Definition:** The amount of "Customer Support" we must give up to gain 1 more unit of "Server Capacity."

**Instructions:**
Use the sliders below to act as the CTO.
1.  **Total Dev Hours:** Represents your Scarcity constraint.
2.  **Allocation to Servers:** Represents your Choice.

Observe how the **Red Dot** moves. The "Loss" in Support output as you move right is your Opportunity Cost.

```python
def plot_interactive_ppf(total_hours, allocation_to_server):
    # 1. Calculate the curve data
    server_hours_range = np.linspace(0, total_hours, 100)
    support_hours_range = total_hours - server_hours_range
    
    server_output_curve = production_function(server_hours_range, efficiency=10)
    support_output_curve = production_function(support_hours_range, efficiency=12)

    # 2. Calculate the specific point chosen by the user
    current_server_hours = allocation_to_server
    current_support_hours = total_hours - current_server_hours
    
    # Handle edge case where allocation > total (should be handled by slider constraints, but good practice)
    if current_server_hours > total_hours:
        current_server_hours = total_hours
        
    current_server_out = production_function(current_server_hours, efficiency=10)
    current_support_out = production_function(current_support_hours, efficiency=12)

    # 3. Plot
    fig, ax = plt.subplots(figsize=(10, 6))
    ax.plot(server_output_curve, support_output_curve, color='navy', linewidth=2, label='PPF Constraint')
    
    # Plot the user's choice
    ax.plot(current_server_out, current_support_out, 'ro', markersize=12, label='Current Strategy', markeredgecolor='white')
    
    # Draw guide lines
    ax.vlines(current_server_out, 0, current_support_out, linestyle=':', color='red')
    ax.hlines(current_support_out, 0, current_server_out, linestyle=':', color='red')

    # Set Limits and Labels
    ax.set_title(f'Resource Allocation Dashboard', fontsize=14)
    ax.set_xlabel(f'Server Capacity (Units): {current_server_out:.1f}', fontsize=12, color='darkred')
    ax.set_ylabel(f'Support Capacity (Units): {current_support_out:.1f}', fontsize=12, color='darkblue')
    
    ax.set_xlim(0, production_function(500, 10) * 1.1) # Fixed axis to prevent jumping
    ax.set_ylim(0, production_function(500, 12) * 1.1)
    ax.legend()
    plt.show()

# Create Widgets
style = {'description_width': 'initial'}
w_total = widgets.IntSlider(min=100, max=500, step=50, value=200, description='Total Hours (Scarcity)', style=style)
w_alloc = widgets.IntSlider(min=0, max=200, step=10, value=100, description='Alloc to Server (Choice)', style=style)

# Link slider max value to total hours
def update_range(*args):
    w_alloc.max = w_total.value
w_total.observe(update_range, 'value')

widgets.interactive(plot_interactive_ppf, total_hours=w_total, allocation_to_server=w_alloc)
```

### **6. Animation: Economic Growth via Technology**

How do we break the constraints of scarcity? In economics, the answer is usually **Technology**.

If your developers get better tools (e.g., AI Coding Assistants, better IDEs), they become more efficient. This implies an increase in the `efficiency` variable in our function.

This shifts the entire PPF curve **outward**. You can produce *more* of both goods without hiring more people. This is **Economic Growth**.

```python
# Setup for Animation
fig, ax = plt.subplots()
plt.close() # Prevent double plotting in notebook output

# Constants for animation
anim_total_hours = 200
anim_server_hours = np.linspace(0, anim_total_hours, 100)
anim_support_hours = anim_total_hours - anim_server_hours

def animate(frame):
    ax.clear()
    
    # Tech Multiplier increases over time (frame)
    # Frame 0 = 1.0x (Baseline), Frame 50 = 2.0x (Double efficiency)
    tech_multiplier = 1.0 + (frame / 50.0) 
    
    # Calculate new curve with better technology
    # We multiply the base efficiency (10 and 12) by the tech_multiplier
    server_output = production_function(anim_server_hours, efficiency=10 * tech_multiplier)
    support_output = production_function(anim_support_hours, efficiency=12 * tech_multiplier)
    
    # Plot curve
    ax.plot(server_output, support_output, color='green', linewidth=2)
    ax.fill_between(server_output, support_output, color='green', alpha=0.1)
    
    # Labels
    ax.set_title(f'Economic Growth Simulation\nTech Multiplier: {tech_multiplier:.2f}x', fontsize=14)
    ax.set_xlabel('Server Capacity')
    ax.set_ylabel('Customer Support')
    
    # Fixed Axis to show expansion clearly
    ax.set_xlim(0, 300) 
    ax.set_ylim(0, 350)
    ax.grid(True)
    ax.text(10, 20, "Technology shifts the curve OUTWARD", fontsize=10)

# Create Animation
anim = animation.FuncAnimation(fig, animate, frames=50, interval=100)

# Display as HTML (JavaScript) for Notebook compatibility
rc('animation', html='jshtml')
anim
```
